# 01 — Data Collection: first look at the raw pulls

Raw Statcast data from `scripts/pull_statcast.py`. Nothing here is filtered or grouped — this is just to see what's there and where the gaps are.

```python
import pandas as pd

pd.set_option('display.max_columns', 200)
pd.set_option('display.max_rows', 200)

FILES = [
    'sasaki_statcast_mlb_2025',
    'sasaki_statcast_mlb_2026',
    'yamamoto_statcast_mlb_2024',
    'yamamoto_statcast_mlb_2025',
    'yamamoto_statcast_mlb_2026',
]
data = {f: pd.read_csv(f'../data/raw/{f}.csv', parse_dates=['game_date']) for f in FILES}
for f, d in data.items():
    print(f'{f}: {d.shape[0]:,} rows x {d.shape[1]} cols, {d.game_date.min().date()} to {d.game_date.max().date()}')
```
